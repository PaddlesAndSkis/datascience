The strategy was to impute values where it could be reasonably guessed what a good value could be and for those that could not, the mean of the Age column was used.

**Age Rules-of-thumb for Imputing:**

The other attributes, Parch and SibSp, can be used to reasonbly estimate a passenger's age where the age is missing.

* If Parch = 0 then the passenger does not have any parents or children.  Therefore, it is assumed the person is an adult as children most likely wouldn't be traveling by themselves.  Use the mean age for adults (>= 18).


* If Parch = 1 and SibSp > 1 then the passenger has a parent or child but has more than one sibling or spouse.  Therefore, it is assumed the person is most likely a child traveling with their family and not an adult with a spouse and a sibling.  Use the mean age for children (< 18).


* If Parch = 2 and SibSp > 1 then the passenger has two parents or children and more than one sibling / spouse.  Similar to above, it is assumed the person is most likely a child traveling with their family and not an adult with two children, a spouse and a sibling.  Use the mean age for children (< 18).


* If Parch > 2 then it is assumed the person is an adult with more than 2 chidren as typically, someone would not have more than two parents.  However, this case does not exist in the dataset.


**Replace With Mean**

Finally, after the above transformations have been made, there are only 7 records remaining without an Age.  It's too difficult to determine a scenario for these as it could be either an adult traveling with children, or a child traveling with adults etc.  Therefore, fill these NaN records with the average age for all passengers.

```python
# Technical Note:
#
# A class, Titanic_Class, has been created with the methods required for the analysis.
# This allows for easier analysis with different attribute scenarios in the cells below.
#
```

```python
# Import the libraries.
#
import pandas as pd
import matplotlib.pyplot as plt


# Define the Class: Titanic_Class
#
class Titanic_Class:
    
    # Constructor
    #
    def __init__ (self, filename):
        
        # Initialize the attributes.
        #
        self.filename = filename
        
        
    # Getter method
    #
    def get_dataframe(self):
        
        # Return the dataframe attribute.
        #
        return self.tdf
       
        
    # load_dataset
    #
    def load_dataset(self):
        
        # Load in the CSV dataset file.
        #
        self.tdf = pd.read_csv(self.filename)
        
        print ("The CSV dataset file,", self.filename, "has been loaded.")
        
        
    # add_columns
    #
    def add_columns(self):
              
        # Add a new column, Age_Group
        #
        self.tdf.loc[self.tdf['Age'] < 18, 'Age_Group'] = 'child'
        self.tdf.loc[(self.tdf['Age'] >= 18) & (self.tdf['Age'] < 40), 'Age_Group'] = 'young adult'
        self.tdf.loc[(self.tdf['Age'] >= 40) & (self.tdf['Age'] < 65), 'Age_Group'] = 'midlife adult'
        self.tdf.loc[self.tdf['Age'] >= 65, 'Age_Group'] = 'senior adult'
        
        # Add a new column, Family
        #
        self.tdf.loc[(self.tdf['SibSp'] == 0) & (self.tdf['Parch'] == 0), 'Family'] = 'single'
        self.tdf.loc[(self.tdf['SibSp'] > 0) | (self.tdf['Parch'] > 0), 'Family'] = 'family' 

        # Add a new column, Parent_Child_Group
        #
        self.tdf.loc[(self.tdf['Parch'] == 0), 'Parent_Child_Group'] = 'no parent or child'
        self.tdf.loc[(self.tdf['Parch'] > 0), 'Parent_Child_Group'] = 'parent or child' 

        # Add a new column, Extended_Family.
        #
        self.tdf.loc[(self.tdf['Parch'] > 0) & (self.tdf['SibSp'] > 0), 'Extended_Family'] = 'strong family'
        self.tdf.loc[(self.tdf['Parch'] > 0) & (self.tdf['SibSp'] == 0), 'Extended_Family'] = 'moderate family'
        self.tdf.loc[(self.tdf['Parch'] == 0) & (self.tdf['SibSp'] > 0), 'Extended_Family'] = 'moderate family'
        self.tdf.loc[(self.tdf['Parch'] == 0) & (self.tdf['SibSp'] == 0), 'Extended_Family'] = 'no family'

        print ("The columns have been loaded.  The new columns are: ", self.tdf.columns)
        
        
    # analyse_dataset
    #
    def analyse_dataset(self, group_args):
        
        # Create a pivot table based on the group arguments (e.g., Pclass, Sex etc.) and
        # calculate the mean.  Sort the values in descending order.
        #
        p_df = self.tdf.pivot_table(index=group_args, values='Survived', aggfunc='mean').sort_values(by='Survived', ascending=False)

        print ('----------------------------')
        print (p_df)
        print ('----------------------------\n')
        
        # Plot a barchart of the dataframe.
        #
        self.plot_barchart(p_df)   

        
    # identify_incomplete_column_data
    #
    def identify_incomplete_column_data(self):
        
        # Print out the columns with null values along with how many records
        # are null for each column.
        #
        print ('----------------------------')
        print (self.tdf.isnull().sum())
        print ('----------------------------')
        
        
    # print_data
    #    
    def print_data(self):
        
        # Print out general information about the data.
        #
        print ('\nSome general information about the dataset:\n')
        print ('----------------------------')
        print ('Total number of passengers: \t', self.tdf.loc[:, 'Sex'].shape[0])
        print ('Male passengers: \t\t', self.tdf[(self.tdf['Sex'] == 'male')].shape[0])
        print ('Female passengers: \t\t', self.tdf[(self.tdf['Sex'] == 'female')].shape[0])
        print ('Mean survival rate: \t\t', self.tdf['Survived'].mean())
        print ('----------------------------')

    
    # deal_with_missing_values
    #
    def deal_with_missing_values(self):

        # Use the child and young adult mean ages to fill records with empty Age values.
        #
        
        # Calculate the child mean age (< 18 yrs).
        #
        cdf = self.tdf[(self.tdf['Age'] < 18)]
        child_mean_age = cdf.loc[:, ['Age']].mean().values[0]
        print ("Child Mean Age = ", child_mean_age)

        # Calculate the young_adult mean age (Age > 18).
        #
        adf = self.tdf[(self.tdf['Age'] >= 18)]
        adult_mean_age = adf.loc[:, ['Age']].mean().values[0]
        print ("Young Adult Mean Age = ", adult_mean_age)
 
        # Underlying assumptions for filling records with empty Age values:
        #
        #    if Parch = 0 then it is assumed the person is an adult
        #                 as children wouldn't be traveling solo; doesn't matter about SibSp
        #    if Parch = 1 and SibSp > 1 then it is assumed the person is a child 
        #                 as the person is traveling with a parent and more than one sibling
        #    if Parch = 2 and SibSp > 1 then it is assumed the person is a child
        #                 as the person is traveling with two parents and more than one sibling
        #    if Parch > 2 then it is assumed the person is an adult
        #                 as typically someone won't have more than 2 parents; 
        #                 however, this is not in the data set
        #
        
        # Replace the age with the adult mean: No Age and Parch = 0 with the adult mean.
        #
        self.tdf.loc[(self.tdf['Age'].isnull()) & (self.tdf['Parch'] == 0), 'Age'] = adult_mean_age
        
        # Replace the age with the adult mean: No Age and Parch > 2 with the adult mean.
        #
        self.tdf.loc[(self.tdf['Age'].isnull()) & (self.tdf['Parch'] > 2), 'Age'] = adult_mean_age

        # Replace the age with the child mean: No Age and Parch = 1 and SibSp > 1 with the child mean.
        #
        self.tdf.loc[(self.tdf['Age'].isnull()) & (self.tdf['Parch'] == 1) & (self.tdf['SibSp'] > 1), 'Age'] = child_mean_age

        # Replace the age with the child mean: No Age and Parch = 2 and SibSp > 0 with the child mean.
        #
        self.tdf.loc[(self.tdf['Age'].isnull()) & (self.tdf['Parch'] == 2) & (self.tdf['SibSp'] > 1), 'Age'] = child_mean_age

        # After the above transformations, there are only 7 records remaining that are empty.
        # For these remaining records, it really can't be determined what age they are as they
        # could be either children or adults.
        # Therefore, just replace the remaining ages with the mean of all ages.
        #
        new_age_mean = self.tdf['Age'].mean()

        self.tdf['Age'] = self.tdf['Age'].fillna(new_age_mean)

        
    # plot_barchart
    #
    def plot_barchart(self, df):
        
        # Plot the dataframe as a bar chart.
        #
        ax = df.plot.bar(rot=0)
```
