## Question #1: Count symbols

(3 points)

In this question, you are given a string `s` which represents a DNA string. The string `s` consists of symbols `'A', 'C', 'G',` and `'T'`. An example of a length 21 DNA string is `"ATGCTTCAGAAAGGTCTTACG."` 

Your task is to write a code which will count the number of times each of the symbols `'A', 'C', 'G'`, and `'T'` occur in `s`. Your code should generate a **list of 4 integers** and **print it out**.

```python
# Here is the DNA string:

s = 'AGCTTTTCATTCTGACTGCAACGGGCAATATGTCTCTGTGTGGATTAAAAAAAGAGTGTCTGATAGCAGC'
```

```python
# Method 1: use a for loop.
#

# Initialize the counter list where "A" = element 0, "C" = element 1, "G" = element 2
# and "T" = element 3.
#
count_list = [0, 0, 0, 0]

# Iterate over the length of the String, s.
#
for i in range(len(s)):

    # Determine if the character is either A, C, G or T.
    #
    if s[i] == "A":
        count_list[0] += 1
    elif s[i] == "C":
        count_list[1] += 1
    elif s[i] == "G":
        count_list[2] += 1
    elif s[i] == "T":
        count_list[3] += 1
    else:
        # Do nothing.
        #
        pass
  
print ("The number of times that [A C G T] appears is", count_list)

print ("\nor...\n")

# Method 2: use the count function
#

# Initalize the dna_symbols and empty dna_symbol_counts.
#
dna_symbols = ["A", "C", "G", "T"]
dna_symbol_counts = []

# Iterate over the dna_symbols.
#
for dna_symbol in dna_symbols:
    
    # Count the number of times the dna symbol appears.
    #
    dna_symbol_counts.append(s.count(dna_symbol))
    
print (dna_symbols, "appears in the DNA string", dna_symbol_counts, "times.")
```

## Question #3: Is a data point within a rectangle?

(4 points)

Write a function `isIn()` which returns **boolean `True`** if a point is within a rectangle specified by two sets of coordinates and **boolean `False`** if the point is outside the rectangle. The function should accept three parameters:
- the first parameter is a set of coordinates which defines one of the corners of the rectangle, 
- the second parameter is also a set of coordinates that defines the second corner,
- the third set of coordinates defines a single point which is being tested.

For example, 
- `isIn((1,2), (3,4), (1.5, 3.2))` should return `True`, 
- `isIn((4,3.5), (2,1), (3, 2))` should return `True`, 
- `isIn((-1,0), (5,5), (6,0))` should return `False`,
- `isIn((4,1), (2,4), (2.5,4.5))` should return `False`.

Test your function with at least 2 different sets of data points in addition to the examples above.

**NOTES:** 
1. If the point being tested is on the side of the rectangle, consider it to be within the rectangle. For example, if the rectangle is defined as `(1,2), (3,4)` and the point is `(2,2)`, the function should return `True`.
2. In this assignment, we assume that the edges of the rectangle are parallel to coordinate axes.
3. We also assume that the first parameter does not always represent the left corner of the rectangle and the second parameter is not always the right corner. The function should work correctly either way. Please note the second test condition above where the first parameter, `(4,3.5)`, represents the top-right corner and the second parameter, `(2,1)`, represents left-bottom corner. 

```python
# isIn
#
# This function tests to see if a point is contained within a
# rectangle given two corner points.
#
# param  firstCorner   : tuple
# param  secondCorner  : tuple
# param  point         : tuple
# return True if the point is contained in the rectangle, False otherwise
#
def isIn(firstCorner=(0,0), secondCorner=(0,0), point=(0,0)):
    
    # Initialize the rectangle orientation identifiers.
    #
    rightRectangle, leftRectangle, upRectangle, downRectangle = False, False, False, False
    
    # Default the return value to False.
    #
    result = False
    
    # Identify the horizontal orientation of the rectangle.
    #
    if firstCorner[0] < secondCorner[0]:
        rightRectangle = True
    else:
        leftRectangle = True
        
    # Identify the vertical orientation of the rectangle.
    #
    if firstCorner[1] < secondCorner[1]:
        upRectangle = True
    else:
        downRectangle = True
    
    # Determine if the point is within the applicable rectangle.  Only identify
    # the success path as the default return value is False.
    #
    if rightRectangle and upRectangle:
    
        # Right and upwards facing rectangle.
        #
        if point[0] >= firstCorner[0] and point[0] <= secondCorner[0]:
            if point[1] >= firstCorner[1] and point[1] <= secondCorner[1]:
                result = True
            
    elif leftRectangle and upRectangle:
    
        # Left and upwards facing rectangle.
        #
        if point[0] <= firstCorner[0] and point[0] >= secondCorner[0]:
            if point[1] >= firstCorner[1] and point[1] <= secondCorner[1]:
                result = True
    
    elif rightRectangle and downRectangle:
    
        # Right and downwards facing rectangle.
        #
        if point[0] >= firstCorner[0] and point[0] <= secondCorner[0]:
            if point[1] <= firstCorner[1] and point[1] >= secondCorner[1]:
                result = True
    
    elif leftRectangle and downRectangle:
    
        # Left and downwards facing rectangle.
        #
        if point[0] <= firstCorner[0] and point[0] >= secondCorner[0]:
            if point[1] <= firstCorner[1] and point[1] >= secondCorner[1]:
                result = True

    else:
        # Do nothing.
        #
        pass
    
    # Return the result (True or False).
    #
    return result
            

# testPoint
#
# A small utility function to format output.
#
# param  firstCorner   : tuple
# param  secondCorner  : tuple
# param  point         : tuple
# return True if the point is contained in the rectangle, False otherwise
#
def testPoint(firstCorner=(0,0), secondCorner=(0,0), point=(0,0)):
    
    print ("Is the point", point, "in a rectangle with corners of", firstCorner, "and", secondCorner,"?")
    print (isIn(firstCorner, secondCorner, point), "\n")

    
# Start of main program.
#

# Test the question examples.
#
testPoint ((1,2), (3,4), (1.5, 3.2)) # True
testPoint((4,3.5), (2,1), (3, 2))    # True
testPoint((-1,0), (5,5), (6,0))      # False
testPoint((4,1), (2,4), (2.5,4.5))   # False

# Test the side.
#
testPoint((1,2), (3,4), (2,2))       # True

# Test other points.
#
testPoint((3,4), (5,6), (2,2))       # False - outside
testPoint((3,4), (5,6), (3,4))       # True - corner
testPoint((3,4), (5,6), (5,6))       # True - corner

# Test defaults.
#
testPoint()                          # True - default to (0,0), (0,0), (0,0)
```
