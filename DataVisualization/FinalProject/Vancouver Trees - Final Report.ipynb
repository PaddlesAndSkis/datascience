#### Data Binning

As there are a lot of columns with a wide range of values, let's bin some of these data fields to facilitate our analysis.  This will create a few more fields of interest.

Let's bin diameter based on quantiles.

```python
min_diameter = van_trees_df['diameter'].min()
q1_diameter = van_trees_df['diameter'].quantile(.25)
q2_diameter = van_trees_df['diameter'].quantile(.5)
q3_diameter = van_trees_df['diameter'].quantile(.75)
max_diameter = van_trees_df['diameter'].max()

van_trees_df.loc[van_trees_df['diameter'] < q1_diameter, 'diameter_group'] = "Small"
van_trees_df.loc[(van_trees_df['diameter'] >= q1_diameter) & (van_trees_df['diameter'] <= q2_diameter), 'diameter_group'] = "Medium"
van_trees_df.loc[(van_trees_df['diameter'] >= q2_diameter) & (van_trees_df['diameter'] <= q3_diameter), 'diameter_group'] = "Large"
van_trees_df.loc[van_trees_df['diameter'] >= q3_diameter, 'diameter_group'] = "Extra Large"
```

Now let's bin the height based on the range.

```python
van_trees_df.loc[van_trees_df['height_range_id'] < 3, 'height_group'] = "Short"
van_trees_df.loc[(van_trees_df['height_range_id'] >= 3) & (van_trees_df['height_range_id'] <= 6), 'height_group'] = "Medium"
van_trees_df.loc[van_trees_df['height_range_id'] >= 7, 'height_group'] = "Tall"
```

#### Imputation

Earlier in the report, it was shown that half the date values were null or unknown.  These records could be removed but that would remove 50% of the data set.  Instead, let's impute that if the year is unknown then it wasn't planted before the first initial year (start of record keeping?).  Therefore, we will just make the year an arbitrary 1980 to represent all dates before those that are known.  These values can always be filtered out during data analysis, if required.

```python
# Create a new Year field based on extracting the Year from the date.

van_trees_df['Year'] = van_trees_df['date_planted'].dt.year

# Impute the null values in this new field with 1980.

van_trees_df['Year'] = van_trees_df['Year'].fillna(value=1980)
```

Let's create a new field, *era*, to bin the Year into decades.

```python
van_trees_df.loc[van_trees_df['Year'] < 1990, 'era'] = "1980s"
van_trees_df.loc[(van_trees_df['Year'] >= 1990) & (van_trees_df['Year'] < 2000), 'era'] = "1990s"
van_trees_df.loc[(van_trees_df['Year'] >= 2000) & (van_trees_df['Year'] < 2010), 'era'] = "2000s"
van_trees_df.loc[van_trees_df['Year'] >= 2010, 'era'] = "2010s"
```

Lastly, let's bin based on longitude to divide Vancouver into east and west.

```python
east_west_mid = (van_trees_df['longitude'].min() + van_trees_df['longitude'].max()) / 2

van_trees_df.loc[van_trees_df['longitude'] < east_west_mid, 'east_west'] = 'west'
van_trees_df.loc[van_trees_df['longitude'] >= east_west_mid, 'east_west'] = 'east'
```

## Preparation

Before we start answering the questions of interest, let's set some common variables and define our helper function.

```python
# Define the numerical columns of that will be useful in our analysis.

numeric_cols = ['diameter', 'height_range_id', 'latitude', 'longitude']

# Get the minimum and maximum values for latitude.

min_lat = van_trees_df['latitude'].min()
max_lat = van_trees_df['latitude'].max()

# Get the minimum and maximum values for longitude.

min_lon = van_trees_df['longitude'].min()
max_lon = van_trees_df['longitude'].max()

# Get the minimum and maximum values for Year.

min_year = van_trees_df['Year'].min()
max_year = van_trees_df['Year'].max()
```

```python
# Dataframe Groupby Columns and Aggregate
#
# C. Turner - Programming in Python for Data Science Course (2022)

# create_dataframe_group_by_columns_and_agg
#
def create_dataframe_group_by_columns_and_agg(df, group_args, agg_column, agg_arg):
    """
    
    Given a dataframe, grouping columns, a column on which to aggregate data
    and the type of aggregate function to invoke, this function creates a sorted 
    data grouping.
    
    Parameters
    ----------
    df : pandas.core.frame.DataFrame
        The dataframe on which to group data
    group_args : list
        The columns to group
    agg_colummn : str
        The column to invoke the aggregate function
    agg_arg : str
        The aggregate function to invoke
        
    Returns
    -------
    pandas.core.frame.DataFrame 
        The grouped dataframe

    Raises
    ------
    
    TypeError
        If the input argument df is not of type DataFrame

    Exception
        If the input argument agg_column does not exist in the dataframe or the
        aggregate function is not valid

    Examples
    --------
    >>> create_dataframe_group_by_columns_and_agg(movie_merge_df, ['genre'], 'movie_title', 'count')
    """
    
    # Ensure that the dataframe parameter is indeed a dataframe.
    
    if type(df) is not pd.core.frame.DataFrame:
        raise TypeError("The first parameter needs to be a dataframe")

    # Verify that the x_axis exists in the dataframe.
    
    if agg_column not in df.columns.tolist():
        raise Exception("Column" + agg_column + "does not exist in the dataframe.")
    
    # Ensure that the specified aggregate function is valid.
    
    agg_arg_list = ['count', 'sum', 'mean', 'max', 'min']
    
    if agg_arg not in agg_arg_list:
        
        raise Exception("Aggregate function", agg_arg,"is not valid.")
    
    # Create a grouping based on the group arguments, aggregate column and the
    # aggregate arguments (e.g., mean, count).
    
    p_df = (df.groupby(by=group_args, as_index=False).agg({agg_column : agg_arg}).
            sort_values(by=agg_column, ascending=False)
            )

    # Return the dataframe.
    
    return p_df
```

## Questions

### Q1. What are the most planted city trees in Vancouver?

To answer this question, a horizontal, side-by-side bar chart is going to be used.  Bar charts are best suited to visualize this data as the results are contained in a single value per group.  As the common names of trees can be long, the bar chart will be horizontal to make the chart easier to read.  Also, only the top 10 trees will be used.


```python
# Group by common_name and genus_name (curb just has to be another column in the dataframe).

common_name_genus_df = create_dataframe_group_by_columns_and_agg(van_trees_df, ['common_name', 'genus_name', 'height_group'], 'curb', 'count')
sorted_common_name_genus_df = common_name_genus_df.sort_values(by='curb', ascending=False)

# Create the chart.

common_name_bar_chart = alt.Chart(sorted_common_name_genus_df.head(10)).mark_bar().encode(
    alt.Y('common_name', title='Common Tree Name', sort='x'),
    alt.X('curb', title='Number of Trees'),
    alt.Color('height_group', title='Height')
).properties(width=400, height=300, title='Most Planted Trees by their Common Name')

# Create the Genus bar chart using the same data as the Common Name bar chart.

common_name_genus_bar_chart = common_name_bar_chart | common_name_bar_chart.encode(alt.Y('genus_name', title='Genus', sort='x')).properties(title='Most Planted Trees by Genus')

common_name_genus_bar_chart
```

Two bar charts are presented side-by-side to visualize both the trees' common names and genus.  The data is sorted such that the larger values are closest to the x-axis.

From the first bar chart, <i>Kwanzan Flowering Cherry</i>, <i>Pissard Plum</i> and <i>Norway Maple</i> are the most common types of trees planted across Vancouver.    

In the second bar chart, <i>Prunus</i> and <i>Acer</i> are the most common genus of trees.

### Q2. What are the most common size of trees planted in Vancouver?

To answer this question, a faceted, scatter chart is going to be used to present the height and diameter by neighbourhood.  The diameter is binned using the assigned category during data pre-processing to make the results more readable.  The sort order of the scatter plot is such that smaller height and diameter trees will appear intuitively in the bottom left-hand side of the x and y axis.  Finally, the size and colour of the data points are sized to the count of trees in that neighbourhood to better visualize the differences.  As the data element is quantitative, a single colour, blue, which gradually becomes darker to represent a larger concentration of trees is used to better illustrate the size of the count (as opposed to using different colours).  The size is also represented in a tooltip for easier comparison.



```python
# Provide a sort order to better illustrate the data.

diameter_order = ['Small', 'Medium', 'Large', 'Extra Large']

# Create the chart.

height_diameter_neighbourhood_scatter = alt.Chart(van_trees_df, width=700, height=400).mark_circle().encode(
    #x='diameter',
    alt.X('diameter_group', title='Diameter (binned)', sort=diameter_order),
    alt.Y('height_range_id', title='Height'),
    alt.Color('count(neighbourhood_name)', title='Number of Trees'),
    size='count(neighbourhood_name)',
    tooltip='count(neighbourhood_name)'
).properties(title='Height vs Diameter', width=150, height=150)

# Facet the chart by neighbourhood.

height_diameter_neighbourhood_scatter_faceted = height_diameter_neighbourhood_scatter.facet('neighbourhood_name', columns=6, title='Heights and Diameter of Trees by Neighbourhood')

height_diameter_neighbourhood_scatter_faceted
```

<p style="text-align: center;"><i>Figure 2: Sizes of Trees by Neighbourhood</i></p>

<b>Results:</b>

The results are presented for each of the 22 neighbourhoods of Vancouver.

From Question 1, we saw that most trees planted are either small or medium height trees.  From our correlation analysis during the EDA process, the height and diameter of the trees had the strongest correlation.  Therefore, we would expect to see most trees planted in each neighbourhood to be small / medium in both height and diameter.

Looking at the data, that is mostly what we see for each of the neighbourhoods - that the most common size of trees planted are small / medium sized trees.  However, Dunbar and Kitsilano do seem to have larger trees planted as part of their neighbourhoods.  Even Kerrisdale and Shaughnessy have a larger number of larger trees in their neighbourhoods.  This could be explained by the fact that these neighbourhoods may have more mature trees than other neighbourhoods.

Renfrew-Collingwood and Kensington-Cedar Cottage, in particular, seem to have a very pronounced number of small trees.



### Q3. Does the city plant trees primarily in certain neighbourhoods or are they distributed equally across the city?

To answer this question, a faceted, scatter plot is used to visualize the distribution of trees planted over the decades in the different neighbourhoods.  A tooltip consisting of the <i>neighbourhood</i>, <i>common name</i>, <i>genus</i> and <i>species name</i> is also included for easier comparison.

```python
# Create the chart.  Don't name the legend as it is self-explanatory from the chart title.

neighbourhood_tree_loc = alt.Chart(van_trees_df).mark_circle().encode(
    alt.X('longitude', title='Longitude', scale=alt.Scale(domain=[min_lon, max_lon])),
    alt.Y('latitude', title='Latitude', scale=alt.Scale(domain=[min_lat, max_lat])),
    alt.Color('neighbourhood_name', title=None),
    tooltip=['neighbourhood_name', 'common_name', 'genus_name', 'species_name']
).properties(title='Tree Distribution by Neighbourhood')

# Facet the scatter plot by era.

neighbourhood_tree_loc_faceted = neighbourhood_tree_loc.facet('era', columns=4, title='Trees Planted by Neighbourhood by Era')

neighbourhood_tree_loc_faceted
```

<p style="text-align: center;"><i>Figure 3: Trees Planted by Neighbourhood by Era</i></p>

<b>Results:</b>

As can be seem from the chart, most trees that were planted in the 1980's (or earlier!) and over the next few decades, were planted all over the city and not just in a few select neighbourhoods.

This is good news from a resident's perspective as it indicates that the city wants to ensure equal planting of trees regardless of the neighbourhood.

The scatter plot using longitude and latitude works well to answer this question rather than a chart that simply indicates the neighbourhood (e.g., bar chart) as this type of chart easily illustrates the distribution.  Faceting by <i>era</i> illustrates that tree planting is distributed evenly over the past few decades.  The colour in the chart is by neighbourhood.  Typically, more than five categories shouldn't use a colour channel (i.e., there are 22 neighourhoods); however, it works in this case as the various colours demarcate the neighbourhoods.

### Q4. Does the city plant a consistent number of trees over the years?

To answer this question, a timeline line chart indicating the year and the number of trees planted is most likely the best visualization.  A tooltip is created to show the Year and the number of trees planted for easier reference.  

To get a better sense of the data distribution, we will filter out the trees where the date was unknown and imputed with a date of 1980.

```python
# Create the line chart.  No need to include the Year x-axis title.

tree_plant_line_chart = alt.Chart(van_trees_df[van_trees_df['Year'] > 1980]).mark_line().encode(
    alt.X('Year', title=None, axis=alt.Axis(format='d')),
    alt.Y('count()', title='Number of Trees Planted')
).properties(height=400, width=400, title='Number of Trees Planted by the City Each Year')

# Add a point for each year to better illustrate the trend.

tree_plant_line_points_chart = tree_plant_line_chart + tree_plant_line_chart.mark_point().encode(tooltip=['Year', 'count()'])

tree_plant_line_points_chart
```

<p style="text-align: center;"><i>Figure 4: Number of Trees Planted by the City Each Year</i></p>

<b>Results:</b>

This is a good chart to visualize the number of trees planted by the city over the past few decades.  Apart from low periods in 1989 and 2016, the answer is yes, the city has been fairly consistent planting trees over the years.  

One thing of note, the dips in the line graph may also be explained by the fact that the data set used is a subset of the entire data set (5,000 records).  Therefore, while there most likely were low points at those times, they may have been less pronounced than what the graph is showing.


### Q5. Are certain trees more prevalent in certain neighbourhoods?

To answer this question, a heat map yields the best results by looking at the genus of the tree by neighbourhood.  The <i>heat</i> of the heatmap will be based on the number of trees to give a clear indication of how many of those trees were planted in that neighbourhood.  This is the value contained in the tooltip to make it easier to see how large the value is for that particular rectangle in the heat map.

First of all, create a new dataframe and view the outcome.  Note that 'curb' is just used as a common field that can be used to store the tree count.

```python
# Group by genus_name and neighbourhood_name (curb just has to be another column in the dataframe).

genus_pd = create_dataframe_group_by_columns_and_agg(van_trees_df, ['genus_name', 'neighbourhood_name'], 'curb', 'count')

genus_pd.head()
```

<p style="text-align: center;"><i>Table 5: Neighbourhood and Top Genus View </i></p>

```python
# To reduce chart 'noise', only focus on those neighbourhoods where there are more than 5 trees planted.

top_genus_pd = genus_pd[genus_pd['curb'] > 5]

genus_neighbourhood_heatmap = alt.Chart(top_genus_pd).mark_rect().encode(
    alt.X('neighbourhood_name', title='Neighbourhood'),
    alt.Y('genus_name', title='Tree Genus'),
    alt.Color('curb', legend=alt.Legend(title='Number of Trees')),
    tooltip=['curb']
).properties(width=700, title=alt.TitleParams("Most Common Trees Planted in the Neighbourhoods by Genus", anchor='middle', subtitle = "Minimum of Five Trees of a Type Genus Planted are Included"))

genus_neighbourhood_heatmap
```

<p style="text-align: center;"><i>Figure 5: Most Common Genus of Trees by Neighbourhood</i></p>

<b>Results:</b>

The results are to be expected based on Question 1.  In that question, the two most popular genus types of trees were <i>Prunus</i> and <i>Acer</i>.  This chart exhibits just that - most trees planted are in the Acer and Prunus genus and are wide spread around Vancouver (i.e., the city has planted a lot of them and they have planted them in all neighbourhoods).  

Curiously, there are some types of trees that can only be found in some neighbourhoods, albeit in small numbers.  Therefore, if you are a fan of trees from the <i>Parrotia</i>, <i>Syringa</i> and <i>Thuja</i> genus, you may want to live in neighbourhoods such as Renfrew-Collingwood or Kensington-Cedar Cottage.

This question is getting a bonus graph.  A faceted, heatmap will illustrate the height range and diameter (size) of the trees planted over the past few decades based on if the neighbourhood is on the east side or west side of Vancouver.  The east or west side of Vancouver is plotted based on the mid-point of the minimum and maximum longitude points.  Also, this categorization was chosen as there are only two categories.  If neighbourhood was used, it would be more than five colours and would be too difficult to distinguish.  As some of the standard colour schemes were making it too difficult to read, a custom colour scheme, Pale Green and Bubblegum Pink (<b>DesignWizard</b>), is used which seems to improve readability.  Furthermore, to add more details, a toolip is added to indicate the neighbourhood.


```python
# Create the chart.  Don't name the legend as it is self-explanatory from the chart title.

neighbourhood_tree_size_heatmap = alt.Chart(van_trees_df).mark_rect(opacity=0.5).encode(
    alt.X('diameter', title='Diameter', type='quantitative', bin=alt.Bin(maxbins=60), scale=alt.Scale(zero=False)),
    alt.Y('height_range_id', title='Height Range', type='quantitative', bin=alt.Bin(maxbins=30), scale=alt.Scale(zero=False)),
    alt.Color('east_west', legend=alt.Legend(title='East or West side of Vancouver'), scale=alt.Scale(range=['#CBCE91FF', '#EA738DFF'])),
    tooltip=['neighbourhood_name']
).properties(title='Tree Size per Neighbourhood')

# Create the faceted heatmap chart by era.

neighbourhood_tree_size_heatmap_faceted = neighbourhood_tree_size_heatmap.facet('era', columns=4, title=alt.TitleParams("Height Range and Diameter of Trees Planted in Vancouver Over the Past Few Decades", anchor='middle', subtitle = "Colour Based on the Neighbourhood being in East or West side of Vancouver"))

neighbourhood_tree_size_heatmap_faceted
```

<p style="text-align: center;"><i>Figure 6: Height and Diameter of Trees by East and West side of Vancouver</i></p>

<b>Results:</b>

From the faceted chart legend, all the neighbourhoods are represented.  This is more visible in the earlier decades.  The most recent decade only shows several neighbourhoods out of the 22.  However, this doesn't make sense given the chart in Question 3, and is most likely explained by binning the diameters.  However, it does show from a high level, that trees are planted in all the various neighbourhoods by the city over the years.

Faceting in Questions 2, 3 and 5 were effective at showing data in different categories (e.g., neighbourhood, era) to be able to compare data without crowding an individual chart.  The use of colour, lightness variation, and size has also helped to visually indicate results to give a quicker, better understanding of data without presenting sometimes difficult-to-interpret data tables.  

There are also times when a simple line chart in Question 4 is sufficient to convey the message as timeline charts are quite common in today's media and are intuitive to understand.  Therefore, when deciding on visual channels in charts, it is sometimes best to use what is commonly presented and understood by the population (e.g., size of points, positioning of data in bar charts and darker shades that draws viewer attention to indicate information such as higher counts).

```python
# Dashboard Panel 2

diameter_click = alt.selection_multi(fields=['diameter_group'])

# Create a 'title-less' diameter group bar chart. 

diameter_group_bar_chart = (alt.Chart(van_trees_df).mark_bar().encode(
    alt.X('count()', title='Number of Trees'),
    alt.Y('diameter_group', title='Diameter Group', sort=diameter_order),
    alt.Color('diameter_group', title=None),
    opacity=alt.condition(diameter_click, alt.value(0.9), alt.value(0.2)))
.add_selection(diameter_click)).properties(width=300)

diameter_group_bar_chart = diameter_group_bar_chart.add_selection(diameter_click)

# Create a scatter chart with the clickable diameter group bar chart.

bars_neighbourhood_chart = alt.Chart(van_trees_df).mark_circle().encode(
    alt.X('longitude', title='Longitude', scale=alt.Scale(domain=[min_lon, max_lon])),
    alt.Y('latitude', title='Latitude', scale=alt.Scale(domain=[min_lat, max_lat])),
    color=alt.condition(diameter_click, 'neighbourhood_name', alt.value('lightgray'), legend=alt.Legend(title=None)),
    tooltip=['neighbourhood_name', 'common_name', 'genus_name', 'species_name']
).properties(title='Tree Distribution by Neighbourhood and Diameter Group').add_selection(diameter_click)

bars_neighbourhood_chart = bars_neighbourhood_chart.encode(opacity=alt.condition(diameter_click, alt.value(0.9), alt.value(0.1)))

# Create Dashboard chart panel #2.

dashboard_panel_2 = bars_neighbourhood_chart | diameter_group_bar_chart
```

```python
# Dashboard Panel 3

# Chart: Neighbourhood tree distribution with clickable legend.

# Create a scatter plot.

click_legend = alt.selection_multi(fields=['neighbourhood_name'], bind='legend')

neighourhood_click_legend_chart = alt.Chart(van_trees_df).mark_circle().encode(
    alt.X('longitude', title='Longitude', scale=alt.Scale(domain=[min_lon, max_lon])),
    alt.Y('latitude', title='Latitude', scale=alt.Scale(domain=[min_lat, max_lat])),
    color=alt.condition(click_legend, 'neighbourhood_name', alt.value('lightgray'), legend=alt.Legend(title=None)),
    tooltip=['neighbourhood_name', 'common_name', 'genus_name', 'species_name']
).properties(title='Tree Distribution by Neighbourhood').add_selection(click_legend)#.interactive()

neighourhood_click_legend_chart = neighourhood_click_legend_chart.encode(opacity=alt.condition(click_legend, alt.value(0.9), alt.value(0.1)))

neighourhood_click_legend_chart = neighourhood_click_legend_chart.add_selection(click_legend).interactive()

# Create Dashboard chart panel #3.

dashboard_panel_3 = neighourhood_click_legend_chart
```

```python
# Dashboard Panel 4

# Chart: Neighbourhood tree distribution with date slider.  No need to include the Year x-axis title.

year_slider = alt.selection_interval()

neighbourhood_date_slider_bar_chart = alt.Chart(van_trees_df[van_trees_df['Year'] > 1980]).mark_bar().encode(
    alt.X('Year', title=None, axis=alt.Axis(format='d')),
    alt.Y('count()', title='Number of Trees'),
    opacity=alt.condition(year_slider, alt.value(0.9), alt.value(0.1))
).properties(height=100).add_selection(year_slider)

# Create the corresponding scatter plot chart.

neighourhood_date_slider_scatter_chart = alt.Chart(van_trees_df).mark_circle().encode(
    alt.X('longitude', title='Longitude', scale=alt.Scale(domain=[min_lon, max_lon])),
    alt.Y('latitude', title='Latitude', scale=alt.Scale(domain=[min_lat, max_lat])),
    alt.Color('neighbourhood_name', legend=alt.Legend(title=None)),
    opacity=alt.condition(year_slider, alt.value(0.9), alt.value(0.1)),
    tooltip=['neighbourhood_name', 'common_name', 'genus_name', 'species_name']
).properties(title='Tree Distribution by Neighbourhood and Year Planted')

# Create Dashboard chart panel #4.

dashboard_panel_4 = neighourhood_date_slider_scatter_chart & neighbourhood_date_slider_bar_chart
```
